```python
from PIL import Image
import pytesseract
import cv2
import os
```

```python
# Explicitly set Tesseract path
pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

# Verify Tesseract path
if not os.path.exists(pytesseract.pytesseract.tesseract_cmd):
    print(f"Error: Tesseract executable not found at {pytesseract.pytesseract.tesseract_cmd}")
    exit(1)
```

```python
def image_to_text(image_path):
    try:
        # Verify image path
        if not os.path.exists(image_path):
            return f"Error: Image not found at {image_path}"
        
        # Load the image
        img = cv2.imread(image_path)
        if img is None:
            return f"Error: Could not load image at {image_path}"
        
        # Convert to grayscale
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        
        # Preprocessing for Bengali text
        gray = cv2.resize(gray, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)
        gray = cv2.GaussianBlur(gray, (3, 3), 0)
        gray = cv2.adaptiveThreshold(
            gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2
        )
        
        # Extract text using Bengali language
        text = pytesseract.image_to_string(gray, lang='ben')
        return text
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
# Example usage
image_path = r"F:\New folder\2.png"
print("Extracted Text:")
print(image_to_text(image_path))
```
